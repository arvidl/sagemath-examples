````python
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic data
np.random.seed(42)
num_timepoints = 6
time_points = [0, 3, 6, 12, 26, 52]

def generate_trajectory(initial_value, final_value, num_points, noise_level):
    x = np.linspace(0, 52, num_points)
    y = initial_value + (final_value - initial_value) * (1 - np.exp(-x / 10))
    noise = np.random.normal(0, noise_level, num_points)
    return y + noise

# Define class parameters
classes = [
    {"name": "Class 1", "color": "red", "initial": 140, "final": 40, "noise": 5, "count": 22},
    {"name": "Class 2", "color": "green", "initial": 100, "final": 50, "noise": 10, "count": 104},
    {"name": "Class 3", "color": "blue", "initial": 120, "final": 80, "noise": 15, "count": 18}
]

# Plot the trajectories
plt.figure(figsize=(10, 6))

for cls in classes:
    for _ in range(cls["count"]):
        trajectory = generate_trajectory(cls["initial"], cls["final"], num_timepoints, cls["noise"])
        plt.plot(time_points, trajectory, color=cls["color"], alpha=0.3)
    
    # Plot mean trajectory
    mean_trajectory = generate_trajectory(cls["initial"], cls["final"], num_timepoints, 0)
    plt.plot(time_points, mean_trajectory, color=cls["color"], linewidth=2, 
             label=f'{cls["name"]} (n={cls["count"]})')

plt.xlabel('Weeks')
plt.ylabel('Psychosis Symptoms')
plt.title('Reduction of psychosis symptoms trajectories (Synthetic Data)')
plt.legend()
plt.show()

# Print sample data
print("Sample data for each class:")
for cls in classes:
    sample_trajectory = generate_trajectory(cls["initial"], cls["final"], num_timepoints, cls["noise"])
    print(f"{cls['name']}: {sample_trajectory}")
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Generate synthetic data
np.random.seed(42)
num_timepoints = 6
time_points = [0, 3, 6, 12, 26, 52]

def generate_trajectory(initial_value, final_value, num_points, noise_level):
    x = np.linspace(0, 52, num_points)
    y = initial_value + (final_value - initial_value) * (1 - np.exp(-x / 10))
    noise = np.random.normal(0, noise_level, num_points)
    return y + noise

# Define class parameters
classes = [
    {"name": "Class 1", "color": "red", "initial": 140, "final": 40, "noise": 5, "count": 22},
    {"name": "Class 2", "color": "green", "initial": 100, "final": 50, "noise": 10, "count": 104},
    {"name": "Class 3", "color": "blue", "initial": 120, "final": 80, "noise": 15, "count": 18}
]

# Generate data
data = []
for cls in classes:
    for _ in range(cls["count"]):
        trajectory = generate_trajectory(cls["initial"], cls["final"], num_timepoints, cls["noise"])
        data.append([cls["name"]] + list(trajectory))

# Create DataFrame
columns = ["Class"] + [f"Week_{t}" for t in time_points]
df = pd.DataFrame(data, columns=columns)

# Plot the trajectories
plt.figure(figsize=(10, 6))

for cls in classes:
    class_data = df[df["Class"] == cls["name"]]
    class_array = class_data.iloc[:, int(1):].values  # Convert Sage Integer to Python int
    for row in class_array:
        plt.plot(time_points, row, color=cls["color"], alpha=0.3)
    
    mean_trajectory = np.mean(class_array, axis=0)
    plt.plot(time_points, mean_trajectory, color=cls["color"], linewidth=2, 
             label=f'{cls["name"]} (n={len(class_data)})')

plt.xlabel('Weeks')
plt.ylabel('Psychosis Symptoms')
plt.title('Reduction of psychosis symptoms trajectories (Synthetic Data)')
plt.legend()
plt.show()

# Print DataFrame info
print(df.info())
print("\nFirst few rows:")
print(df.head())
````

### Linear Mixed Models -  Children’s Growth Example
http://campus.murraystate.edu/academic/faculty/cmecklin/STA565/_book/an-introduction-to-linear-mixed-models.html

The traditional analytic techniques for handling correlated data are repeated measures ANOVA and mixed models. The linear mixed model for correlated data can be stated (in a regression model format) as:

$$
y=\beta_0+(\beta_1 x_1 + \beta_2 x_2 + \cdots + \beta_p x_p) +(b_0 + b_1 z_1 + b_2 z_2 + \cdots + b_q z_q) + \epsilon
$$

where the $x$ variables represent the fixed effects and the $z$ variables represent the random effects. We can express the linear mixed model in matrix form:

$$
\boldsymbol{Y}_i=\boldsymbol{X \beta} + \boldsymbol{Zb} + \boldsymbol{\epsilon}
$$

where  $\boldsymbol{Y}_i$ is the  n × 1 response vector for the  ith response,  X is the  n × p design matrix for the  p fixed effects, β is the  p × 1 parameter vector for fixed effects,  Z is the  n × q design matrix for the  
q random effects, and b is the  q × 1 random effects parameter vector. 

Unlike the traditional linear model, which is typically fit with **least squares**, the linear mixed model is either fit with **maximum likelihood** or with REML, **restricted maximum likelihood**. REML is a variation of maximum likelihood that generally has less bias in estimates of variability.

Mixed models are well suited for clustered data, repeated measures, longitudinal data, and hierarchical models. While methods based on classical ANOVA can handle certain special cases well (such as repeated measures ANOVA from a balanced design with no missing data), mixed models are crucial for handling more complex situations, including missing data, individuals measured as different time periods, etc.

Mixed models can also help us avoid the statistical sin of **pseudoreplication**, a source of error in statistical inference where we treat the data as being independent when it is not. This leads to us exaggerating the size of the sample and thus the degrees of freedom and  $p$-value, which can lead to erroneously concluding statistical significance that does not actually exist (i.e. Type I error). Psuedoreplication typically occurs either in an observational study with a hierarchical structure or a designed experiment with differing spatial and/or temporal scales.

These linear mixed models are very useful for dealing with situations where there is heterogeneity in response to stimuli and treatments. This often occurs in biological and medical problems, and mixed models methodology allows us to introduce multilevel random effects to accurately model between cluster and within cluster variation. These models can be implemented in R with the lmer function from the lme4 package, or PROC MIXED in SAS.




The **children’s weight** example reads in a file in the “long” format It consists of weight measurements of Asian children in a British community on up to five occasions, at irregular time intervals, with the child’s age at that measurement given in years, weight in kg. Each individual child ($id$) will be treated as a random effect; there are 198 children in this data set.

````python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy import stats

# Download the data (if not already done)
url = "http://campus.murraystate.edu/academic/faculty/cmecklin/Child.txt"
child_wt = pd.read_csv(url, delimiter="\s+")

# Create a quadratic term for age
child_wt['age_squared'] = child_wt['age'] ** 2

# Function to fit model and return log-likelihood
def fit_model(formula):
    model = smf.mixedlm(formula, data=child_wt, groups=child_wt["id"])
    results = model.fit()
    return results

# Fit full model and nested models
full_model = fit_model("weight ~ age + age_squared")
model_without_age_squared = fit_model("weight ~ age")
model_without_age = fit_model("weight ~ age_squared")
null_model = fit_model("weight ~ 1")

# Function to perform likelihood ratio test
def lr_test(full_model, reduced_model):
    lr_stat = -2 * (reduced_model.llf - full_model.llf)
    df = len(full_model.params) - len(reduced_model.params)
    p_value = stats.chi2.sf(lr_stat, df)
    return lr_stat, df, p_value

# Perform likelihood ratio tests
print("ANOVA-like table for fixed effects:")
print("Term\t\tChi-square\tDf\tp-value")

# Test for age_squared
lr_age_squared, df_age_squared, p_age_squared = lr_test(full_model, model_without_age_squared)
print(f"age_squared\t{lr_age_squared:.4f}\t\t{df_age_squared}\t{p_age_squared:.4f}")

# Test for age
lr_age, df_age, p_age = lr_test(full_model, model_without_age)
print(f"age\t\t{lr_age:.4f}\t\t{df_age}\t{p_age:.4f}")

# Test for both age terms together
lr_both, df_both, p_both = lr_test(full_model, null_model)
print(f"age + age_squared\t{lr_both:.4f}\t\t{df_both}\t{p_both:.4f}")
````

## A predictive model


We aim to construct and evaluate a predictive model where 
"CE and NE volume correlation [i.e. exam-specific CE/NE ratio, or CE / (CE+NE) ratio ?] in repeated MRI scans will be analyzed, to assess differences in radiological treatment responses. 
This includes two radiological tumor expressions (CE + NE), in 3D-volume, over time. Patients will be predicted to 1 of 3 groups of treatment response prospectively, that will ultimately be compared to ground truth measures ..."

Reformulated, this involves:
- (i) a multiclass outcome $y = \{G_1, G_2, G_3\}$ denoting treatment response groups  
- (ii) observed data $\mathbf{X} = \{x_{ij}\}$ where $i=1,...,n$  and $j=0,...,T-1$   such that 
$x_{i\cdot} = x_{iCE}^{(t=0)}, x_{iNE}^{(t=0)}, ..., x_{iCE}^{(t=T-1)}, x_{iNE}^{(t=T-1)}$
where $n$  is number of patients, $T$ is number of timepoints (i.e. repeated MRI scans in the longitudinal study), and $x_{iCE}^{(t=p)}, x_{iNE}^{(t=p)}$ is the CE and NE volumes respectively in patient $i$ at time point $p$ (values can be missing if $i$ has no scan at time point $p$, assuming $\mathbf{X}$ is a regular $n \times 2T$ matrix), and
- (iii) a prediction model $f$ with (learned) parameters $\theta$, i.e.

$$
y \sim f(\mathbf{X}, \theta)
$$

Putting this into a predictive (machine learning) framework, even with simple models (e.g. multinomial logistic regression), given the sparsity of the data, one could consider a leave-one-out cross-validation to assess the goodness and performance of the model.
In the assessment, one could consider confusion matrices, balanced accuracy and ROC-AUC.

Leave-One-Out Cross-Validation (LOOCV) is a special case of k-fold cross-validation where \( k \) is equal to the number of observations in the dataset (in our case: k=n). 
In each iteration, one observation is used as the test set, and the remaining observations are used as the training set.

Here's a step-by-step guide to performing LOOCV using a multinomial logistic regression in Python with the `LogisticRegression` class from `scikit-learn`:

````python
#%pip install scikit-learn
````

````python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import accuracy_score
````

````python
data = load_iris()
X, y = data.data, data.target
````

````python
model = LogisticRegression(multi_class='multinomial', solver='lbfgs', max_iter=1000)
loo = LeaveOneOut()
````

````python
accuracies = []

for train_index, test_index in loo.split(X):
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]
   
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracies.append(accuracy_score(y_test, y_pred))

average_accuracy = np.mean(accuracies)
print(f"Average Accuracy: {average_accuracy:.2f}")
````

### Import libraries

We import several libraries that we will use throughout the code. NumPy and Pandas are used for data manipulation, Matplotlib is used for data visualization, Scikit-learn is used for machine learning, and we import specific functions like load_iris, train_test_split, DecisionTreeClassifier, confusion_matrix, and plot_tree from the respective Scikit-learn modules.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix
from sklearn.tree import plot_tree
````

### Load the Iris dataset

````python
iris = load_iris()
````

### Split the dataset into training and testing sets

````python
# X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=float(0.2), random_state=42)
````

### Train a decision tree classifier model

````python
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)
````

### Evaluate the model on the testing set

````python
score = model.score(X_test, y_test)
print(f"Accuracy: {score:.2f}")
````

### Compute the confusion matrix

````python
y_pred = model.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
````

### Plot the confusion matrix

````python
fig, ax = plt.subplots()
im = ax.imshow(cm, interpolation="nearest", cmap=plt.cm.Blues)
ax.figure.colorbar(im, ax=ax)
ax.set(xticks=np.arange(cm.shape[1]),
       yticks=np.arange(cm.shape[0]),
       xticklabels=iris.target_names, yticklabels=iris.target_names,
       title="Confusion matrix",
       xlabel="Predicted label",
       ylabel="True label")
plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(j, i, format(cm[i, j], "d"), ha="center", va="center",
                color="white" if cm[i, j] > cm.max() / 2 else "black")
fig.tight_layout()
````
